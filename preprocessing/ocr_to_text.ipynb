# Preprocessing 1: OCR correction

**LanguageTool** is an open-source grammar tool, also known as the spellchecker for OpenOffice. This library allows you to detect grammar errors and spelling mistakes through a Python script.

! pip install language-tool-python

```python
from pathlib import Path
import language_tool_python
import os
```

```python
tool = language_tool_python.LanguageTool('it')
```

```python
# Correzione OCR con LanguageTool a blocchi per evitare crash
def correggi_ocr_language_tool_blocchi(testo: str, dimensione_blocco=5000) -> str:
    """
    Correzione OCR con LanguageTool a blocchi per evitare crash
    """
    print("Correzione OCR a blocchi con LanguageTool...")

    blocchi = [
        testo[i:i + dimensione_blocco]
        for i in range(0, len(testo), dimensione_blocco)
    ]

    risultato = []

    for i, blocco in enumerate(blocchi, 1):
        print(f"   Blocco {i}/{len(blocchi)}")
        corretto = tool.correct(blocco)
        risultato.append(corretto)

    return "".join(risultato)
```

```python
# brginning from a specific file of the folder
for file in os.listdir("../year_clean/"):

    if file > "un_1970":
    
        print(f"File: {file} ", end="")
        
        filename = os.fsdecode(file)

        output_file = "../ocr_clean/" + filename + '_clean.txt'

        with open("../year_clean/" + filename, "r", encoding="utf-8", errors="ignore") as f:
            testo = f.read()

        testo_corretto = correggi_ocr_language_tool_blocchi(testo)

        with open(output_file, "w", encoding="utf-8") as f:
            f.write(testo_corretto)

        print("OK   ")
```
