```python
from pathlib import Path
import language_tool_python

tool = language_tool_python.LanguageTool('it')


def correggi_ocr_language_tool_blocchi(testo: str, dimensione_blocco=5000) -> str:
    """
    Correzione OCR con LanguageTool a blocchi per evitare crash
    """
    print("Correzione OCR a blocchi con LanguageTool...")

    blocchi = [
        testo[i:i + dimensione_blocco]
        for i in range(0, len(testo), dimensione_blocco)
    ]

    risultato = []

    for i, blocco in enumerate(blocchi, 1):
        print(f"   Blocco {i}/{len(blocchi)}")
        corretto = tool.correct(blocco)
        risultato.append(corretto)

    return "".join(risultato)
```

```python
from pathlib import Path

input_file = Path("../year_clean/un_1948")
output_file = Path("../year_clean/un_1948_clean.txt")

with open(input_file, "r", encoding="utf-8", errors="ignore") as f:
    testo = f.read()


testo_fasttext = correggi_ocr_language_tool_blocchi(testo)

with open(output_file, "w", encoding="utf-8") as f:
    f.write(testo_fasttext)

print("OK")
```
